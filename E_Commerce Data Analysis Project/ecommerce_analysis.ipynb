#### QUESTIONS.

##### 1. Which Products generates the most revenue

```python
products_revenue = df.groupby('Product')['Revenue'].sum().sort_values(ascending = False).reset_index()
products_revenue.style.format({'Revenue':'${:,.0f}'}).hide( axis = 'index')
```

```python
products_revenue.plot(kind = 'barh', x = 'Product', y = 'Revenue', legend = False, figsize = (9,5))
plt.gca().invert_yaxis()
plt.ylabel("Products")
plt.xlabel("Revenue")
plt.title('Revenue By Products')
plt.show()
```

##### 2. Which Categories generates the most revenue

```python
categories_revenue = df.groupby('Category')['Revenue'].sum().reset_index().sort_values(by = 'Revenue',ascending = False)
categories_revenue.style.format({'Revenue':'${:,.0f}'})
```

```python
categories_revenue.plot(kind = 'bar', x = 'Category', y = 'Revenue', legend = False, figsize = (10,5))
plt.xlabel("Categories")
plt.ylabel("Revenue")
plt.title('Revenue By Categories')
plt.xticks(rotation = 0)
plt.show()
```

##### 3. Which Region has the highest numbers of orders?

```python
region_counts = df['Region'].value_counts()
region_counts
```

```python
region_counts.plot(kind = 'bar')
plt.xlabel("Region")
plt.ylabel("Orders")
plt.title('Orders by Region')
plt.xticks(rotation = 0)
plt.show()
```

```python
sns.boxplot(y=df['Revenue'])
plt.title("Distribution of Order Values")
plt.ylabel("Revenue")
plt.show()
```
